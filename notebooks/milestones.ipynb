# Milestones gantt diagram

```python
import textwrap

import matplotlib.pyplot as plt

# Weeks
WEEKS = 14

# Tasks: (label, start, end)
TASKS_IMPL = [
    ("Reference run (original code)", 0, 1),
    ("ST model implementation", 2, 4),
    ("Baseline reproduction", 5, 6),
    ("DrugProt experiment (RE)", 7, 8),
    ("SciArg experiment (AM)", 9, 10),
]
TASKS_WRITE = [
    ("Lit. & dataset review", 0, 1),
    ("Background & Related Work", 2, 4),
    ("Methods", 5, 8),
    ("Results & Discussion", 9, 11),
    ("Final Revision", 12, 13),
]

# Milestones: (label, week)
MILESTONES = [("M1", 2), ("M2", 5), ("M3", 9), ("M4", 12), ("Final", 14)]


# Helper for wrapped text
def wrap_text(text, width=16):
    return "\n".join(textwrap.wrap(text, width))


# Plot
fig, ax = plt.subplots(figsize=(12, 5))
bar_h = 0.6
y_impl, y_write = 2, 1

# Plot Implementation tasks
for task, s, e in TASKS_IMPL:
    ax.barh(
        y_impl, e - s + 1, left=s, height=bar_h, align="center", color="skyblue", edgecolor="black"
    )
    ax.text(s + (e - s + 1) / 2, y_impl, wrap_text(task), va="center", ha="center", fontsize=9)

# Plot Writing tasks
for task, s, e in TASKS_WRITE:
    ax.barh(
        y_write,
        e - s + 1,
        left=s,
        height=bar_h,
        align="center",
        color="lightgreen",
        edgecolor="black",
    )
    ax.text(s + (e - s + 1) / 2, y_write, wrap_text(task), va="center", ha="center", fontsize=9)

# Milestones
for label, week in MILESTONES:
    ax.axvline(x=week, color="red", linestyle="--", linewidth=1)
    ax.text(
        week,
        y_impl + 0.4,
        label,
        color="red",
        ha="center",
        va="bottom",
        fontsize=9,
        fontweight="bold",
    )

# Formatting
ax.set_yticks([y_impl, y_write])
ax.set_yticklabels(["Implementation", "Writing"])
ax.set_xlim(0, WEEKS)  # boundaries

# Primary ticks at week boundaries (0..14), no labels
ax.set_xticks(range(WEEKS + 1))
ax.set_xticklabels([])

# Secondary ticks in the middle (0.5..13.5) with labels
ax.set_xticks([i + 0.5 for i in range(WEEKS)], minor=True)
ax.set_xticklabels([str(i + 1) for i in range(WEEKS)], minor=True)
ax.tick_params(axis="x", which="minor", length=0)

ax.set_xlabel("Weeks")
ax.set_title("Thesis Timeline", pad=25)

plt.tight_layout()
plt.show()

data_path = "./data/"
filename = "thesis_timeline_v4"
fig.savefig(data_path + filename + ".png", dpi=220)
fig.savefig(data_path + filename + ".pdf")
```
